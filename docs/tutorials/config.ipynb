# **Config Files**

This notebook is designed to explain how config files are designed, what role they play, and go through each parameter they contain. We go through an example of an 80k second batch from late July 2025. MARS3 did not register data for this period, so we exclude it from the list. 

## Parameters

First, let's define things that will not change between batches, and that are not impacted by choices in analysis. These are required for all analysis, and form the basis of config files.

```python
ant_names_all = [
    "MARS1",
    "MARS2",
    "MARS3",
    "MARS4",
    "MARS5",
    "MARS6",
    "MARS7",
    "MARS8"]

ant_coords_all = [
    [79.41716147,	-90.76723869,	187.9577],	#MARS1
    [79.41719805,	-90.75873919,	183.0684],	#MARS2		
    [79.41540905,	-90.77299123,	180.4816],	#MARS3			
    [79.38845641,	-91.01920296,	25.1938],	#MARS4
    [79.41830257,	-90.66739545,	59.6242],	#MARS5			
    [79.39798424,	-90.79984241,	41.699],	#MARS6			
    [79.41147412,	-90.69526613,	31.6314],	#MARS7				
    [79.44375769,	-90.71820263,	414.9131]]	#MARS8

ant_paths_all = [
    ["/scratch/mohanagr/summer_2025/baseband/mars1"],
    ["/scratch/mohanagr/summer_2025/baseband/mars2"],
    ["/scratch/mohanagr/summer_2025/baseband/mars3"],
    ["/scratch/mohanagr/summer_2025/baseband/mars4"],
    ["/scratch/mohanagr/summer_2025/baseband/mars5"],
    ["/scratch/mohanagr/summer_2025/baseband/mars6"],
    ["/scratch/mohanagr/summer_2025/baseband/mars7"],
    ["/scratch/mohanagr/summer_2025/baseband/mars8"]]
```

Next, there are batch-specific parameters, which may or may not be needed depending on what analysis is being performed.

```python
# You always need a start and end time.
batch_start_ts = 1753200150
batch_end_ts = 1753286410

# You always need to specify the channels
# Note that these are indices as a fraction of the PFB channel width,
# and may differ from the file's contained channel indices
start_channel = 1834
end_channel =  1852

# Clock offsets are different for each batch.
# They determined by the satellite detection script.
# Therefore, despite being essential for fine timing, they are not needed for 'satdet'
clock_offsets_all = [
    0,         #MARS1 is the reference
    -1884333,  #MARS2
    -2427320,  #MARS4
    -1895548,  #MARS5
    -901873,   #MARS6
    -2123503,  #MARS7
    -2365055   #MARS8
]
```

Finally, there are analysis-dependent parameters. These may be the same between batches, but are dependent on the exact way in which you are treating your data.

```python
# This is the accumulation length (in number of baseband spectra)
# that we use for coarse cross-correlation chunks in satellite detection
coarse_acclen = 3000000

# This is the oversampling factor that we use when re-PFBing 
# data into narrower channels for the UTC fitting and fine timing
osamp = 64

# This is the accumulation length that we use (in re-PFBed spectra)
# when getting visibilities for fine timing and UTC offsets
new_acclen = 1024

# This is the total number of baseband spectra 
# that go into one re-PFBed visibility
pfb_size = 65536   #aka osamp*new_acclen

# This is the accumulation length (in number of baseband spectra)
# that we use when we make visibilities in satellite detection.
# (it's not used much any more)
vis_acclen = 30000
```

## Construct the Dictionary

The dictionary `file` will get saved to `.json` in this exact format.

```python
file = {}

antennas = {}
nant = len(ant_names_all)

for j in range(nant):
    antennas[ant_names_all[j]] = {}
    antennas[ant_names_all[j]]["name"] = ant_names_all[j]
    antennas[ant_names_all[j]]["path"] = ant_paths_all[j]
    antennas[ant_names_all[j]]["coordinates"] = list(ant_coords_all[j])

file['antennas'] = antennas

file["correlation"] = {
    "start_timestamp": batch_start_ts,
    "end_timestamp": batch_end_ts,
    "vis_acclen": vis_acclen,
    "coarse_acclen": coarse_acclen,
    "osamp": osamp,
    "pfb_size": pfb_size,
    "new_acclen": new_acclen
}

file["frequency"] = {
    "start_channel": start_channel,
    "end_channel": end_channel
}
```
