```python
from matplotlib import pyplot as plt
import numpy as np
import sys
from os import path
import os

sys.path.insert(0, path.expanduser("~"))
from albatros_analysis.src.utils import pfb_utils as pu
import cupy as cp
import time
```

```python
#load the data

with np.load("./data/spectra_1830_1840_4096_24000_0.05_clock_1overf_5e-9.npz") as f:
    spec1 = f["spectra1"]
    spec2 = f["spectra2"]
    delays = f["delays"]
```

```python
#up channelize the data
nant = 1
npol = 2
osamp = 16
pfb_size = 65536
channels = np.arange(1830, 1840)
lblock = 4096
cutsize = 10
read_size = pfb_size - 2 * cutsize
timestream_size = read_size * lblock
nchan = len(channels)
new_channels = np.arange(osamp) + channels[:, None] * osamp
new_channels = new_channels.ravel()
new_nchan = len(new_channels)
# first we do regular IPFB that has full band
ipfb1 = pu.StreamingIPFB(
    nant,
    npol,
    channels,
    nblock=pfb_size,
    lblock=4096,
    ntap=4,
    window="hamming",
    cut=cutsize,
)
fpfb1 = pu.StreamingPFB(
    nant, npol, timestream_size=timestream_size, lblock=lblock * osamp
)
```

```python
# first we do regular IPFB that has full band
ipfb2 = pu.StreamingIPFB_IQ(
    nant,
    npol,
    channels,
    nblock=pfb_size,
    lblock=4096,
    ntap=4,
    window="hamming",
    cut=cutsize,
)
new_lblock = ipfb2.lblock
print("new lblock is", new_lblock)
fpfb2 = pu.StreamingPFB(
    nant, npol, timestream_size=timestream_size, lblock=lblock * osamp
)
```
